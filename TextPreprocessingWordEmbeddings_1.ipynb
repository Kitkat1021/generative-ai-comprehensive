## Bag of Words

```python
# Import libraries for numerical computations and data manipulation
# numpy: numerical arrays and mathematical operations
# pandas: DataFrame for structured data (rows, columns) to store text and labels
import numpy as np
import pandas as pd
```

```python
# Create a sample DataFrame with text documents and binary sentiment labels
# 'text' column: contains documents (space-separated words)
# 'output' column: 1 = positive/relevant, 0 = negative/irrelevant (for classification)
df = pd.DataFrame({"text":["people watch stanford",
                         "stanford watch stanford",
                         "people write comment",
                          "stanford write comment"],"output":[1,1,0,0]})

df
```

```python
# CountVectorizer converts text documents into a matrix of token counts
# Each row represents a document, each column represents a unique word (token)
# Values in the matrix are the frequency of each word in each document
# This is the foundation of the Bag of Words (BoW) representation
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
```

```python
# fit_transform() learns the vocabulary from the text data and transforms it simultaneously
# fit(): builds a dictionary of all unique words found in the corpus
# transform(): converts each document into a sparse matrix (optimized for memory usage)
# Result is a sparse matrix where element [i,j] = count of word j in document i
bow = cv.fit_transform(df['text'])
```

```python
# Print the vocabulary dictionary learned from the corpus
# Format: {word: column_index} — each unique word maps to a column in the BoW matrix
# This mapping is used to convert text to numerical vectors for machine learning models
print(cv.vocabulary_)
```

```python
# Convert the sparse matrix to a dense array for visualization
# Each row = one document, each column = one word from the vocabulary
# Each cell = count of how many times that word appears in that document
# For example: row 0 has "comment": 0, "people": 1, "stanford": 1, "watch": 1, "write": 0
bow.toarray()
```
