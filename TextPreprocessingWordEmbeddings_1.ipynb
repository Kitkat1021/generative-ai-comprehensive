## Bag of Words

```python
# Import libraries for numerical computations and data manipulation
# numpy: numerical arrays and mathematical operations
# pandas: DataFrame for structured data (rows, columns) to store text and labels
import numpy as np
import pandas as pd
```

```python
# Create a sample DataFrame with text documents and binary sentiment labels
# 'text' column: contains documents (space-separated words)
# 'output' column: 1 = positive/relevant, 0 = negative/irrelevant (for classification)
df = pd.DataFrame({"text":["people watch stanford",
                         "stanford watch stanford",
                         "people write comment",
                          "stanford write comment"],"output":[1,1,0,0]})

df
```

```python
# CountVectorizer converts text documents into a matrix of token counts
# Each row represents a document, each column represents a unique word (token)
# Values in the matrix are the frequency of each word in each document
# This is the foundation of the Bag of Words (BoW) representation
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
```

```python
# fit_transform() learns the vocabulary from the text data and transforms it simultaneously
# fit(): builds a dictionary of all unique words found in the corpus
# transform(): converts each document into a sparse matrix (optimized for memory usage)
# Result is a sparse matrix where element [i,j] = count of word j in document i
bow = cv.fit_transform(df['text'])
```

```python
# Print the vocabulary dictionary learned from the corpus
# Format: {word: column_index} — each unique word maps to a column in the BoW matrix
# This mapping is used to convert text to numerical vectors for machine learning models
print(cv.vocabulary_)
```

```python
# Convert the sparse matrix to a dense array for visualization
# Each row = one document, each column = one word from the vocabulary
# Each cell = count of how many times that word appears in that document
# For example: row 0 has "comment": 0, "people": 1, "stanford": 1, "watch": 1, "write": 0
bow.toarray()
```

## N grams

```python
df = pd.DataFrame({"text":["people watch stanford",
                         "stanford watch stanford",
                         "people write comment",
                          "stanford write comment"],"output":[1,1,0,0]})

df
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# N-grams capture sequences of n words — useful for capturing word relationships and context
# ngram_range=(2,2) means use only Bigrams: pairs of consecutive words (e.g., "people watch", "watch stanford")
# Bigrams preserve word order and proximity information that Unigrams (single words) lose
cv = CountVectorizer(ngram_range=(2,2))
```

```python
# Apply fit_transform to create Bigram representation — learns vocabulary of word pairs from the corpus
# Each document is now represented as a vector of Bigram frequencies instead of individual word frequencies
# This captures local word context and semantic relationships better than unigrams alone
bow = cv.fit_transform(df['text'])
```

```python
# Print the Bigram vocabulary — shows all unique word pairs and their column indices
# Notice: each entry is now a 2-word sequence (e.g., "people watch", "watch stanford")
# instead of single words, enabling feature vectors to capture word relationships
print(cv.vocabulary_)
```

```python
# ngram_range=(3,3) creates Trigrams: sequences of 3 consecutive words
# Trigrams capture even longer context and are useful for detecting phrase-level patterns
# Example trigrams: "people watch stanford", "stanford watch stanford"
# Higher n-grams are more specific but result in sparser matrices and may overfit
cv2 = CountVectorizer(ngram_range=(3,3))
bow2 = cv2.fit_transform(df['text'])
```

```python
# Print Trigram vocabulary — shows all 3-word sequences found in the corpus
# Trigrams are more specific than Bigrams but result in fewer matches per document
# Useful for capturing semantic phrases (e.g., "watch stanford" vs just "watch")
# Trade-off: higher n-grams are more specific but sparsity increases with corpus size
print(cv2.vocabulary_)
```

## TF - IDF (Term Frequency - Inverse Document Frequency)

```python
df = pd.DataFrame({"text":["people watch stanford",
                         "stanford watch stanford",
                         "people write comment",
                          "stanford write comment"],"output":[1,1,0,0]})

df
```

```python
# TfidfVectorizer converts text documents to TF-IDF (Term Frequency-Inverse Document Frequency) vectors
# TF-IDF weights word importance: frequent words in a document get higher scores, but words that appear in many documents get lower scores
# This avoids the problem with CountVectorizer where common words (like "the", "a") dominate despite carrying little meaning
from sklearn.feature_extraction.text import TfidfVectorizer
tfid = TfidfVectorizer()
```

```python
# fit_transform learns TF-IDF weights from the corpus and applies them to create weighted vectors
# Values are normalized (0 to 1) — words appearing in many documents have lower weights (near 0)
# Rare discriminative words get higher weights, making documents more distinguishable for ML models
# toarray() converts the sparse matrix to a dense 2D array for easier inspection and visualization
arr = tfid.fit_transform(df['text']).toarray()
arr
```

```python
# IDF (Inverse Document Frequency) values for each word in the vocabulary
# Higher IDF = word appears in fewer documents (more discriminative, higher importance)
# Lower IDF = word appears in many documents (less discriminative, appears as 0 in many rows of TF-IDF matrix)
# Formula: IDF(word) = log(total_documents / documents_containing_word)
print(tfid.idf_)
```
